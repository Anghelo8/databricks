```python
%sql
-- 1. Ranking de provincias por sismos
SELECT departamento, provincia, SUM(TRY_CAST(igp_n_sismos AS DOUBLE)) AS total_sismos
FROM big_data_team1.anghelo.integracion_gold
GROUP BY departamento, provincia
ORDER BY total_sismos DESC
LIMIT 10;

-- 2. Ranking de provincias por emergencias INDECI
SELECT departamento, provincia, SUM(TRY_CAST(indeci_n_emergencias AS DOUBLE)) AS total_emergencias
FROM big_data_team1.anghelo.integracion_gold
GROUP BY departamento, provincia
ORDER BY total_emergencias DESC
LIMIT 10;

-- 3. Comparativa de temperaturas por departamento
SELECT departamento,
       ROUND(AVG(TRY_CAST(temp_prom_c AS DOUBLE)), 2) AS temp_promedio,
       ROUND(AVG(TRY_CAST(temp_max_c AS DOUBLE)), 2) AS temp_maxima,
       ROUND(AVG(TRY_CAST(temp_min_c AS DOUBLE)), 2) AS temp_minima
FROM big_data_team1.anghelo.integracion_gold
GROUP BY departamento
ORDER BY temp_promedio DESC;

-- 4. Filtro y ranking: magnitud máxima por provincia
SELECT departamento, provincia, MAX(TRY_CAST(igp_magnitud_max AS DOUBLE)) AS magnitud_maxima
FROM big_data_team1.anghelo.integracion_gold
WHERE TRY_CAST(igp_magnitud_max AS DOUBLE) IS NOT NULL
GROUP BY departamento, provincia
ORDER BY magnitud_maxima DESC
LIMIT 10;

-- 5. Comparativa de precipitación por departamento
SELECT departamento,
       ROUND(AVG(TRY_CAST(precipitacion_mm AS DOUBLE)), 2) AS precipitacion_promedio,
       ROUND(MAX(TRY_CAST(precipitacion_mm AS DOUBLE)), 2) AS precipitacion_maxima
FROM big_data_team1.anghelo.integracion_gold
GROUP BY departamento
ORDER BY precipitacion_promedio DESC;
```
